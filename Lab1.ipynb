```python
import simpy
def cloud_task(env, name, vm, execution_time):
  print(f"{name} arrived at time {env.now}")
  with vm.request() as request:
    yield request
    print(f"{name} started execution at time {env.now}")
    yield env.timeout(execution_time)
    print(f"{name} completed at time {env.now}")

env = simpy.Environment()
vm = simpy.Resource(env, capacity=1)
env.process(cloud_task(env, "Cloudlet-1", vm, 4))
env.process(cloud_task(env, "Cloudlet-2", vm, 3))
env.process(cloud_task(env, "Cloudlet-3", vm, 5))
env.run()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

tasks = [
    ("T1", 10),
    ("T2", 4),
    ("T3", 6),
    ("T4", 8),
    ("T5", 2)
]

waiting_time = []
turnaround_time = []
completion_time = []
current_time = 0

for task, burst in tasks:
  waiting_time.append(current_time)
  current_time += burst
  completion_time.append(current_time)
  turnaround_time.append(current_time)

result = pd.DataFrame({
    "Task": [task for task, burst in tasks],
    "Burst Time": [burst for task, burst in tasks],
    "Waiting Time": waiting_time,
    "Completion Time": completion_time,
    "Turnaround Time": turnaround_time
})

print("=" * 60)
print("FCFS SCHEDULING RESULTS")
print("=" * 60)
print(result)

avg_wait = result["Waiting Time"].mean()
avg_turn = result["Turnaround Time"].mean()
print("\nAverage Waiting Time    :", round(avg_wait, 2))
print("Average Turnaround Time :", round(avg_turn, 2))
print("Makespan                :", completion_time[-1])

fig, ax = plt.subplots(figsize=(10, 2.5))

start = 0
colors = ['skyblue', 'lightgreen', 'orange', 'pink', 'yellow']
for i, (task, burst) in enumerate(tasks):
  ax.broken_barh([(start, burst)], (10, 8), facecolors=colors[i % len(colors)])
  ax.text(start + burst/2,
          14,
          task,
          ha='center',
          va='center',
          fontsize=10,
          fontweight='bold')
  start += burst
ax.set_ylim(5, 25)
ax.set_xlim(0, completion_time[-1])
ax.set_xlabel("Time")
ax.set_ylabel("CPU")
ax.set_title("Gantt Chart - FCFS Scheduling")
ax.set_yticks([])
ax.grid(True)
print("\n")
plt.show()
print("\n")

plt.figure(figsize=(7,4))
plt.bar(result["Task"], result["Waiting Time"])
plt.title("Waiting Time of Tasks")
plt.xlabel("Tasks")
plt.ylabel("Waiting Time")
plt.grid(axis='y')
plt.show()
print("\n")

plt.figure(figsize=(7,4))
plt.bar(result["Task"], result["Turnaround Time"])
plt.title("Turnaround Time of Tasks")
plt.xlabel("Tasks")
plt.ylabel("Turnaround Time")
plt.grid(axis='y')
plt.show()
print("\nProgram Executed Successfully.")
```

```python
tasks = [
    ("T1", 0, 10),
    ("T2", 1, 4),
    ("T3", 2, 6),
    ("T4", 3, 8),
    ("T5", 4, 2)]

waiting = []
turnaround = []
completion = []
response = []

time = 0

for task, arrival, burst in tasks:
  if time < arrival:
    time = arrival
  wt = time - arrival
  rt = wt
  waiting.append(wt)
  response.append(rt)
  time += burst
  completion.append(time)
  turnaround.append(time - arrival)

df = pd.DataFrame({
    "Task":[t[0] for t in tasks],
    "Arrival Time":[t[1] for t in tasks],
    "Burst Time":[t[2] for t in tasks],
    "Waiting Time":waiting,
    "Response Time":response,
    "Completion Time":completion,
    "Turnaround Time":turnaround
})

print("\n")
print("="*75)
print("FIRST COME FIRST SERVE (FCFS) SCHEDULING")
print("="*75)

print(df.to_string(index=False))
avg_wait = df["Waiting Time"].mean()
avg_turn = df["Turnaround Time"].mean()
avg_response = df["Response Time"].mean()
makespan = completion[-1]
cpu_busy = sum(df["Burst Time"])
cpu_util = cpu_busy/makespan*100
throughput = len(tasks)/makespan
print("\nPerformance Metrics")
print("-"*40)
print(f"Average Waiting Time     : {avg_wait:.2f}")
print(f"Average Turnaround Time  : {avg_turn:.2f}")
print(f"Average Response Time    : {avg_response:.2f}")
print(f"Makespan                : {makespan}")
print(f"CPU Utilization          : {cpu_util:.2f}%")
print(f"Throughput              : {throughput:.3f} Tasks/Unit Time")

fig, ax = plt.subplots(figsize=(12,2.5))
colors=['skyblue','lightgreen','orange','pink','yellow']
start=0
for i,row in df.iterrows():
  burst=row["Burst Time"]
  ax.broken_barh([(start,burst)],(10,8), facecolors=colors[i%5], edgecolor='black')
  ax.text(start+burst/2,
          14,
          row["Task"],
          ha='center',
          va='center',
          fontsize=10)
  start+=burst
ax.set_xlim(0,makespan)
ax.set_ylim(5,25)
ax.set_xlabel("Time")
ax.set_yticks([])
ax.set_title("Gantt Chart - FCFS Scheduling")
ax.grid(True)
print("\n")
plt.show()
print("\n")

plt.figure(figsize=(7,4))
plt.bar(df["Task"],df["Waiting Time"])
plt.title("Waiting Time")
plt.xlabel("Task")
plt.ylabel("Time")
plt.grid(axis='y')
plt.show()
print("\n")

plt.figure(figsize=(7,4))
plt.bar(df["Task"],df["Turnaround Time"])
plt.title("Turnaround Time")
plt.xlabel("Task")
plt.ylabel("Time")
plt.grid(axis='y')
plt.show()
print("\nFCFS Scheduling Completed Successfully.")
```

```python
tasks = [
    ("T1", 10),
    ("T2", 4),
    ("T3", 6),
    ("T4", 8),
    ("T5", 2)]

quantum = 2
n = len(tasks)
task_names = [task[0] for task in tasks]
burst_time = [task[1] for task in tasks]
remaining_time = burst_time.copy()
completion_time = [0] * n
waiting_time = [0] * n
turnaround_time = [0] * n
time = 0
gantt = []
while True:
  done = True
  for i in range(n):
    if remaining_time[i] > 0:
      done = False
      start = time
      if remaining_time[i] > quantum:
        time += quantum
        remaining_time[i] -= quantum
        gantt.append((task_names[i], start, time))
      else:
        time += remaining_time[i]
        gantt.append((task_names[i], start, time))
        completion_time[i] = time
        turnaround_time[i] = completion_time[i]
        waiting_time[i] = turnaround_time[i] - burst_time[i]
        remaining_time[i] = 0
  if done:
        break

print("=" * 75)
print("        ROUND ROBIN SCHEDULING")
print("=" * 75)

print("{:<8}{:<12}{:<15}{:<15}{:<15}".format(
    "Task",
    "Burst",
    "Waiting",
    "Turnaround",
    "Completion"))

for i in range(n):
  print("{:<8}{:<12}{:<15}{:<15}{:<15}".format(
        task_names[i],
        burst_time[i],
        waiting_time[i],
        turnaround_time[i],
        completion_time[i]))

avg_wait = sum(waiting_time) / n
avg_turn = sum(turnaround_time) / n
makespan = max(completion_time)
cpu_util = (sum(burst_time) / makespan) * 100
throughput = n / makespan

print("\nPerformance Metrics")
print("-" * 40)
print("Average Waiting Time    :", round(avg_wait, 2))

print("Average Turnaround Time :", round(avg_turn, 2))
print("Makespan               :", makespan)
print("CPU Utilization        :", round(cpu_util, 2), "%")
print("Throughput             :", round(throughput, 3), "Tasks/Unit Time")

fig, ax = plt.subplots(figsize=(12, 2.5))

colors = ['skyblue', 'lightgreen', 'orange', 'pink', 'yellow']

for i, item in enumerate(gantt):
  task, start, end = item
  ax.broken_barh([(start, end-start)],
                   (10, 8),
                   facecolors=colors[task_names.index(task) % 5],
                   edgecolor='black')
  ax.text((start+end)/2,
            14,
            task,
            ha='center',
            va='center',
            fontsize=8)
ax.set_xlim(0, makespan)
ax.set_ylim(5, 25)
ax.set_xlabel("Time")
ax.set_yticks([])
ax.set_title("Round Robin Gantt Chart")
ax.grid(True)
print("\n")
plt.show()
print("\n")

plt.figure(figsize=(7,4))
plt.bar(task_names, waiting_time)
plt.title("Waiting Time")
plt.xlabel("Tasks")
plt.ylabel("Time")
plt.grid(axis='y')
plt.show()
print("\n")

plt.figure(figsize=(7,4))
plt.bar(task_names, turnaround_time)
plt.title("Turnaround Time")
plt.xlabel("Tasks")
plt.ylabel("Time")
plt.grid(axis='y')
plt.show()
print("\nRound Robin Scheduling Completed Successfully.")
```

```python
tasks = [
    ("T1", 10),
    ("T2", 4),
    ("T3", 6),
    ("T4", 8),
    ("T5", 2)]

tasks = sorted(tasks, key=lambda x: x[1])
task_names = []
burst_times = []
waiting_times = []
turnaround_times = []
completion_times = []
current_time = 0
for task, burst in tasks:
  task_names.append(task)
  burst_times.append(burst)
  waiting_times.append(current_time)
  current_time += burst
  completion_times.append(current_time)
  turnaround_times.append(current_time)

print("=" * 75)
print("SHORTEST JOB FIRST (SJF) SCHEDULING")
print("=" * 75)

print("{:<8}{:<12}{:<15}{:<15}{:<15}".format(
    "Task",
    "Burst",
    "Waiting",
    "Turnaround",
    "Completion"))

for i in range(len(tasks)):
  print("{:<8}{:<12}{:<15}{:<15}{:<15}".format(
      task_names[i],
      burst_times[i],
      waiting_times[i],
      turnaround_times[i],
      completion_times[i]))

avg_wait = sum(waiting_times) / len(tasks)
avg_turn = sum(turnaround_times) / len(tasks)
makespan = completion_times[-1]
cpu_util = (sum(burst_times) / makespan) * 100
throughput = len(tasks) / makespan
print("\nPerformance Metrics")
print("-" * 40)
print("Average Waiting Time    :", round(avg_wait, 2))
print("Average Turnaround Time :", round(avg_turn, 2))
print("Makespan              :", makespan)
print("CPU Utilization        :", round(cpu_util, 2), "%")
print("Throughput            :", round(throughput, 3), "Tasks/Unit Time")

fig, ax = plt.subplots(figsize=(10, 2.5))
colors = ['skyblue', 'lightgreen', 'orange', 'pink', 'yellow']
start = 0

for i in range(len(tasks)):
   burst = burst_times[i]
   ax.broken_barh(
        [(start, burst)],
        (10, 8),
        facecolors=colors[i % 5],
        edgecolor='black')
   ax.text(
        start + burst / 2,
        14,
        task_names[i],
        ha='center',
        va='center',
        fontsize=10)
   start += burst

ax.set_xlim(0, makespan)
ax.set_ylim(5, 25)
ax.set_xlabel("Time")
ax.set_yticks([])
ax.set_title("Gantt Chart - SJF Scheduling")
ax.grid(True)
print("\n")
plt.show()
print("\n")

plt.figure(figsize=(7,4))
plt.bar(task_names, waiting_times)
plt.title("Waiting Time")
plt.xlabel("Task")
plt.ylabel("Time")
plt.grid(axis='y')
plt.show()
print("\n")

plt.figure(figsize=(7,4))
plt.bar(task_names, turnaround_times)
plt.title("Turnaround Time")
plt.xlabel("Task")
plt.ylabel("Time")
plt.grid(axis='y')
plt.show()
print("\nSJF Scheduling Completed Successfully.")
```

```python
tasks = ["T1", "T2", "T3", "T4", "T5"]
vms = ["VM1", "VM2", "VM3"]

execution_time = [
    [14, 10, 12],
    [16, 20, 18],
    [9,  12, 10],
    [7,  8,  6],
    [5,  4,  7]]

machine_ready = [0] * len(vms)
remaining_tasks = list(range(len(tasks)))
schedule = []

while remaining_tasks:
  minimum_completion = []
  for t in remaining_tasks:
    completion = []
    for vm in range(len(vms)):
      completion.append(machine_ready[vm] + execution_time[t][vm])
    min_time = min(completion)
    vm_index = completion.index(min_time)
    minimum_completion.append((t, vm_index, min_time))
  selected = max(minimum_completion, key=lambda x: x[2])
  task = selected[0]
  vm = selected[1]
  finish = selected[2]
  start = machine_ready[vm]
  machine_ready[vm] = finish
  schedule.append((tasks[task], vms[vm], start, finish))
  remaining_tasks.remove(task)


print("=" * 75)
print("MAX-MIN TASK SCHEDULING")
print("=" * 75)

print("{:<8}{:<8}{:<12}{:<12}".format(
    "Task", "VM", "Start", "Finish"))
for s in schedule:
  print("{:<8}{:<8}{:<12}{:<12}".format(
      s[0], s[1], s[2], s[3]))

makespan = max(machine_ready)
print("\nPerformance Metrics")
print("-" * 40)
print("Makespan :", makespan)

for i in range(len(vms)):
    print(vms[i], "Completion Time :", machine_ready[i])

fig, ax = plt.subplots(figsize=(10,4))
colors = ['skyblue', 'lightgreen', 'orange', 'pink', 'yellow']
for i, item in enumerate(schedule):
  task, vm, start, finish = item
  vm_index = vms.index(vm)
  ax.broken_barh(
        [(start, finish-start)],
        (vm_index*10+5, 8),
        facecolors=colors[i % len(colors)],
        edgecolor='black'
    )
  ax.text(
        start + (finish-start)/2,
        vm_index*10+9,
        task,
        ha='center',
        va='center',
        fontsize=9)

ax.set_ylim(0, len(vms)*10+5)
ax.set_xlim(0, makespan+5)
ax.set_xlabel("Time")
ax.set_ylabel("Virtual Machines")
ax.set_yticks([9,19,29])
ax.set_yticklabels(vms)
ax.set_title("Gantt Chart - Max-Min Scheduling")
ax.grid(True)
print("\n")
plt.show()
print("\n")

plt.figure(figsize=(6,4))
plt.bar(vms, machine_ready)
plt.title("Completion Time of Virtual Machines")
plt.xlabel("Virtual Machines")
plt.ylabel("Time")
plt.grid(axis='y')
plt.show()
print("\nMax-Min Scheduling Completed Successfully.")
```
