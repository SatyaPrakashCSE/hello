```python
import simpy
import random
import math

# Objective function
# We want to minimize f(x) = x^2
def objective_function(x):
    return x ** 2


# Simulated Annealing process
def simulated_annealing(env, initial_x, initial_temperature,
                        cooling_rate, iterations):

    current_x = initial_x
    current_cost = objective_function(current_x)

    best_x = current_x
    best_cost = current_cost

    temperature = initial_temperature

    print("\nSimulated Annealing Started")
    print("--------------------------------")

    for i in range(iterations):

        # Generate a neighboring solution
        new_x = current_x + random.uniform(-1, 1)
        new_cost = objective_function(new_x)

        # Difference in cost
        delta = new_cost - current_cost

        # Accept the new solution if it is better
        # or with a probability based on temperature
        if delta < 0:
            accept = True
        else:
            probability = math.exp(-delta / temperature)
            accept = random.random() < probability

        if accept:
            current_x = new_x
            current_cost = new_cost

        # Update best solution
        if current_cost < best_cost:
            best_x = current_x
            best_cost = current_cost

        print(
            f"Iteration: {i + 1:3d} | "
            f"x = {current_x:8.4f} | "
            f"Cost = {current_cost:8.4f} | "
            f"Temperature = {temperature:8.4f}"
        )

        # Cool down the temperature
        temperature *= cooling_rate

        # SimPy process waits for one simulation time unit
        yield env.timeout(1)

    print("--------------------------------")
    print("Simulated Annealing Completed")
    print(f"Best solution x = {best_x:.4f}")
    print(f"Minimum value = {best_cost:.4f}")


# Create SimPy environment
env = simpy.Environment()

# Parameters
initial_x = 10
initial_temperature = 100
cooling_rate = 0.95
iterations = 50

# Start the Simulated Annealing process
env.process(
    simulated_annealing(
        env,
        initial_x,
        initial_temperature,
        cooling_rate,
        iterations
    )
)

# Run simulation
env.run()
```

```python
def objective_function(x):
    return -(x - 5) ** 2 + 25


def hill_climbing(start_x, step_size=1):

    current_x = start_x
    current_value = objective_function(current_x)

    while True:

        # Generate neighboring solutions
        left_x = current_x - step_size
        right_x = current_x + step_size

        left_value = objective_function(left_x)
        right_value = objective_function(right_x)

        # Find the best neighbor
        if left_value > current_value:
            next_x = left_x
            next_value = left_value

        elif right_value > current_value:
            next_x = right_x
            next_value = right_value

        else:
            # No better neighbor exists
            break

        # Move to the better neighbor
        current_x = next_x
        current_value = next_value

        print(
            f"x = {current_x}, "
            f"f(x) = {current_value}"
        )

    return current_x, current_value


# Starting solution
start_x = 0

# Apply Hill Climbing
best_x, best_value = hill_climbing(start_x)

print("\nFinal Solution")
print("--------------------")
print("Best x =", best_x)
print("Maximum value =", best_value)
```

```python
import heapq


def best_first_search(graph, heuristic, start, goal):

    # Priority queue: (heuristic value, node)
    priority_queue = []

    heapq.heappush(
        priority_queue,
        (heuristic[start], start)
    )

    visited = set()

    while priority_queue:

        # Select node with the smallest heuristic value
        h_value, current = heapq.heappop(priority_queue)

        if current in visited:
            continue

        print("Visiting:", current)

        visited.add(current)

        # Check whether goal is reached
        if current == goal:
            print("Goal reached:", goal)
            return

        # Add unvisited neighboring nodes
        for neighbor in graph[current]:

            if neighbor not in visited:
                heapq.heappush(
                    priority_queue,
                    (heuristic[neighbor], neighbor)
                )

    print("Goal not found")


# Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}


# Heuristic values
heuristic = {
    'A': 6,
    'B': 4,
    'C': 3,
    'D': 6,
    'E': 2,
    'F': 1,
    'G': 0
}


# Starting and goal nodes
start = 'A'
goal = 'G'


# Execute Best-First Search
best_first_search(graph, heuristic, start, goal)
```
